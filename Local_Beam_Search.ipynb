## Introduction

## Local Beam Search

Local Beam Search is a heuristic search algorithm that keeps track of k best states (beam width) at each step. It starts with k random states, generates all their successors, evaluates them using a heuristic, and selects the top k to continue. This process repeats until the goal is found or a limit is reached. By exploring multiple paths in parallel, it reduces the risk of getting stuck in local optima compared to single-path methods like hill climbing.

## Stochastic Beam Search

Stochastic Beam Search is a variation of Local Beam Search that selects the next beam probabilistically rather than deterministically. Instead of always choosing the top k states with the best heuristic scores, it selects k states based on a probability distribution—states with better scores have a higher chance, but not a guarantee, of being chosen.

## Local vs Stochastic Beam Search

Local Beam Search → Always picks the top k best successors (greedy).

Stochastic Beam Search → Picks k successors randomly, weighted by their heuristic scores, introducing diversity and helping escape local optima.

### Limitations of **Local Beam Search**:

1. **Local Optima**: Since it always selects the best `k` states, it can get stuck in **local optima**, failing to find the global solution.
2. **Limited Exploration**: It may miss out on less obvious solutions due to its deterministic nature, focusing only on the top `k` candidates.
3. **Heuristic Dependence**: The performance heavily relies on the **quality of the heuristic function**; a poor heuristic can lead to suboptimal paths.


### Limitations of **Stochastic Beam Search**:

1. **Randomness**: The random selection introduces **uncertainty**, and there’s no guarantee that it will find the optimal solution.
2. **Inefficiency**: While it may escape local optima, it could also waste resources exploring paths that don’t lead to better solutions.
3. **Heuristic Sensitivity**: Like Local Beam Search, it still relies on the heuristic, and a poor heuristic can impact performance.

Both approaches are prone to **suboptimal solutions** if the search space is large or the heuristic is weak.


## Local Beam Search 

### N-Queens Solver using Local Beam Search

```python
import random

def initial_state(n):
    # * Generate a random initial state for the board
    return [random.randint(0, n-1) for _ in range(n)]

def calculate_attacks(board):
    # * Calculate the number of attacking pairs of queens on the board
    n = len(board)
    attacks = 0
    for i in range(n):
        for j in range(i+1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                attacks += 1
    return attacks

def generate_children(parent, n):
    # * Generate all possible child states by moving one queen at a time
    children = []
    for i in range(n):
        child = list(parent)
        # * Pick a new random position for the queen at index i
        new_position = random.randint(0, n-1)
        while child[i] == new_position:  # * Ensure queen doesn't stay in the same place
            new_position = random.randint(0, n-1)
        child[i] = new_position
        children.append(child)
    return children

def local_beam_search(n, beam_width, max_iterations):
    # * Perform Local Beam Search to find a solution to the N-Queens problem
    # * Initialize beam with random states
    current_states = [initial_state(n) for _ in range(beam_width)]
    
    for _ in range(max_iterations):
        next_states = []
        
        # * Generate children for each state in the current beam
        for state in current_states:
            if calculate_attacks(state) == 0:
                return state  # * Goal state found
            children = generate_children(state, n)
            next_states.extend(children)
        
        # * Sort by the number of attacks and select the best 'beam_width' states
        next_states.sort(key=lambda x: calculate_attacks(x))
        current_states = next_states[:beam_width]
    
    # * If no solution found within max_iterations, return the best state found so far
    return current_states[0]

def print_board(board):
    # * Print the board in a human-readable format
    n = len(board)
    for i in range(n):
        row = ['Q' if j == board[i] else '.' for j in range(n)]
        print(' '.join(row))


# * --------------------------------------------
# * Example usage
n = 8  # * Number of queens (size of the board)
beam_width = 5  # * Beam width parameter
max_iterations = 1000  # * Maximum number of iterations

solution = local_beam_search(n, beam_width, max_iterations)
print("Final solution:")
print_board(solution)
```

### 8-Puzzle Solver using Local Beam Search

```python
import copy

# * Define initial and goal states
initial_state = [
    ['1', '2', '3'],
    ['4', '5', '6'],
    ['7', '_', '8']
]

goal_state = [
    ['1', '2', '3'],
    ['_', '4', '6'],
    ['7', '5', '8']
]

N = 3  # * 3x3 puzzle

# * Heuristic: Manhattan Distance
def manhattan_distance(state):
    total = 0
    for i in range(N):
        for j in range(N):
            tile = state[i][j]
            if tile == '_':
                continue
            for x in range(N):
                for y in range(N):
                    if goal_state[x][y] == tile:
                        total += abs(i - x) + abs(j - y)
    return total

# * Get position of blank tile
def find_blank(state):
    for i in range(N):
        for j in range(N):
            if state[i][j] == '_':
                return i, j

# * Generate all valid children from a state by sliding a tile into the blank
def generate_children(state):
    children = []
    x, y = find_blank(state)
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]  # * up, down, left, right

    for dx, dy in moves:
        nx, ny = x + dx, y + dy
        if 0 <= nx < N and 0 <= ny < N:
            child = copy.deepcopy(state)
            child[x][y], child[nx][ny] = child[nx][ny], child[x][y]
            children.append(child)
    return children

# * Local Beam Search algorithm
def local_beam_search(beam_width, max_iterations):
    current_states = [initial_state]

    for _ in range(max_iterations):
        next_states = []

        for state in current_states:
            if state == goal_state:
                return state
            next_states.extend(generate_children(state))

        # * Remove duplicates based on state string representation
        unique_next_states = []
        seen = set()
        for state in next_states:
            key = str(state)
            if key not in seen:
                seen.add(key)
                unique_next_states.append(state)

        unique_next_states.sort(key=manhattan_distance)
        current_states = unique_next_states[:beam_width]

    return current_states[0]  # * Return best state found if goal not reached

# * Print state in 3x3 format
def print_board(state):
    for row in state:
        print(" ".join(row))
    print()


# * ------------------------------------------
# * Example usage
beam_width = 3
max_iterations = 50

solution = local_beam_search(beam_width, max_iterations)
print("Final solution:")
print_board(solution)
```

## Stochastic Beam Search

### N-Queens Solver using Stochastic Beam Search

```python
import random

def initial_state(n):
    # Generate a random initial state for the board
    return [random.randint(0, n-1) for _ in range(n)]

def calculate_attacks(board):
    # Calculate the number of attacking pairs of queens on the board
    n = len(board)
    attacks = 0
    for i in range(n):
        for j in range(i+1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                attacks += 1
    return attacks

def generate_children(parent, n):
    # Generate all possible child states by moving one queen at a time
    children = []
    for i in range(n):
        child = list(parent)
        new_position = random.randint(0, n-1)
        while child[i] == new_position:
            new_position = random.randint(0, n-1)
        child[i] = new_position
        children.append(child)
    return children

def stochastic_beam_search(n, beam_width, max_iterations):
    # Initialize beam with random states
    current_states = [initial_state(n) for _ in range(beam_width)]
    
    for _ in range(max_iterations):
        next_states = []

        for state in current_states:
            if calculate_attacks(state) == 0:
                return state  # Goal state found
            children = generate_children(state, n)
            next_states.extend(children)

        # Compute fitness scores (inverse of number of attacks)
        fitness_scores = []
        for state in next_states:
            attacks = calculate_attacks(state)
            score = 1 / (1 + attacks)  # Add 1 to avoid division by zero
            fitness_scores.append(score)

        # Normalize to probabilities
        total = sum(fitness_scores)
        probabilities = [score / total for score in fitness_scores]

        # Sample new beam based on fitness probabilities
        current_states = random.choices(next_states, weights=probabilities, k=beam_width)

    # Return best state found if no goal was found
    return min(current_states, key=calculate_attacks)

def print_board(board):
    n = len(board)
    for i in range(n):
        row = ['Q' if j == board[i] else '.' for j in range(n)]
        print(' '.join(row))

# Example usage
n = 8
beam_width = 5
max_iterations = 1000

solution = stochastic_beam_search(n, beam_width, max_iterations)
print("Final solution:")
print_board(solution)
```

### 8-Puzzle Solver using Stochastic Beam Search

```python
import copy
import random

# * Define initial and goal states
initial_state = [
    ['1', '2', '3'],
    ['4', '5', '6'],
    ['7', '_', '8']
]

goal_state = [
    ['1', '2', '3'],
    ['_', '4', '6'],
    ['7', '5', '8']
]

N = 3  # * 3x3 puzzle

# * Manhattan Distance Heuristic
def manhattan_distance(state):
    total = 0
    for i in range(N):
        for j in range(N):
            tile = state[i][j]
            if tile == '_':
                continue
            for x in range(N):
                for y in range(N):
                    if goal_state[x][y] == tile:
                        total += abs(i - x) + abs(j - y)
    return total

# * Find blank tile
def find_blank(state):
    for i in range(N):
        for j in range(N):
            if state[i][j] == '_':
                return i, j

# * Generate all valid children
def generate_children(state):
    children = []
    x, y = find_blank(state)
    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]  # * up, down, left, right

    for dx, dy in moves:
        nx, ny = x + dx, y + dy
        if 0 <= nx < N and 0 <= ny < N:
            child = copy.deepcopy(state)
            child[x][y], child[nx][ny] = child[nx][ny], child[x][y]
            children.append(child)
    return children

# * Stochastic Beam Search algorithm
def stochastic_beam_search(beam_width, max_iterations):
    current_states = [initial_state]

    for _ in range(max_iterations):
        next_states = []

        for state in current_states:
            if state == goal_state:
                return state
            next_states.extend(generate_children(state))

        # * Remove duplicates
        unique_next_states = []
        seen = set()
        for state in next_states:
            key = str(state)
            if key not in seen:
                seen.add(key)
                unique_next_states.append(state)

        # * Compute fitness (inverse of Manhattan distance)
        fitness_scores = []
        for state in unique_next_states:
            h = manhattan_distance(state)
            score = 1 / (1 + h)  # * Add 1 to avoid divide-by-zero
            fitness_scores.append(score)

        # * Normalize scores to probabilities
        total = sum(fitness_scores)
        if total == 0:
            probabilities = [1/len(fitness_scores)] * len(fitness_scores)
        else:
            probabilities = [score / total for score in fitness_scores]

        # * Randomly sample next beam
        if len(unique_next_states) < beam_width:
            current_states = unique_next_states
        else:
            current_states = random.choices(unique_next_states, weights=probabilities, k=beam_width)

    # * Return best found
    return min(current_states, key=manhattan_distance)

# * Print board
def print_board(state):
    for row in state:
        print(" ".join(row))
    print()

# * ------------------------------------------
# * Example usage
beam_width = 3
max_iterations = 50

solution = stochastic_beam_search(beam_width, max_iterations)
print("Final solution:")
print_board(solution)
```
